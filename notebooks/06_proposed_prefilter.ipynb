**Workflow step: Step 4 (a) of Figure 1 — the proposed approach, first variant.**
All JET-flagged entries are put in one list and ranked by an unsupervised anomaly score, so that an auditor with a budget of k entries reviews the most suspicious JET entries first (RQ1).

**Goal.** (1) Build the JET union list (notebook 01). (2) Rank it with each detector and both encodings — variant B: detector fitted on the whole year, scores used inside the list (main); variant A: detector fitted on the JET list only (D2; Cut List item 3). (3) Compare with JET-only and H1 at the budgets of D14. (4) Record which anomalies the prefilter cannot reach (the 93.9 % recall ceiling; input for RQ3).

**Input:** `meta.parquet`, encodings, settings from notebooks 03 / 05. **Output:** `results/06_proposed_prefilter.csv` (overwritten on each run). **Week:** 4 (7–13 Dec).
